```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
import seaborn as sns
```

```python
# load dataset
df = pd.read_csv('../datasets/emails.csv')
df.head()
```

```python
# dataset info
df.info()
```

```python
# statistical summary
df.describe()
```

```python
df.head()
```

```python
# dropping unnecessary features.
df.drop('Email No.', axis=1, inplace=True)
```

```python
# data preparation
X = df.drop("Prediction", axis=1)
y = df['Prediction']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8, random_state=42)
```

```python
# features scaling
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=500) # reducing to top 500 components
X_train = pca.fit_transform(X_train)
X_test = pca.transform(X_test)
```

```python
# getting the classification models
svc = SVC(C=1.0, kernel='rbf', gamma='auto')
knn = KNeighborsClassifier(n_neighbors=7)
```

```python
# traning the models
svc.fit(X_train, y_train)
knn.fit(X_train, y_train)
```

```python
# prediction
svc_pred = svc.predict(X_test)
knn_pred = knn.predict(X_test)
```

```python
# getting the results.
result = {
    "SVC": {
        "Accuracy": accuracy_score(y_test, svc_pred),
        "Precision": precision_score(y_test, svc_pred),
        "Recall": recall_score(y_test, svc_pred),
        "F1": f1_score(y_test, svc_pred)
    },
    "KNN": {
        "Accuracy": accuracy_score(y_test, knn_pred),
        "Precision": precision_score(y_test, knn_pred),
        "Recall": recall_score(y_test, knn_pred),
        "F1": f1_score(y_test, knn_pred)
    }
}
result
```

```python
# confusion matrices
print("Confusion Matrix for SVM")
display(confusion_matrix(y_test, svc_pred))

print("Confusion Matrix for KNN")
display(confusion_matrix(y_test, knn_pred))
```
